## Anatra maintenance programme, November to December 2026

We are carrying out a series of changes to Anatra to bring the operating system up to date and to improve job scheduling and resource utilisation across the cluster. The work starts in the week of 16 November 2026 and will continue until the end of the year.

Where a change causes downtime we will notify all researchers using Anatra by email with approximately two weeks' notice.

The schedule below covers the changes confirmed so far. We will keep it updated as further approvals are put in place, so we advise you to revisit this page for the latest information.

### Adjusting DefMemPerCPU and MaxMemPerCPU

<p>
  <span style="display:inline-block; background:#1f4e79; color:#fff; padding:3px 10px; border-radius:4px; font-weight:600;">16 November 2026</span>
  &nbsp; <strong>Downtime:</strong> None expected
</p>

**Issue:** The current memory-per-CPU default and cap values do not align with the memory available on the compute nodes. This can lead to CPUs being left idle when jobs do not request memory, jobs being allocated more CPUs than they actually use, and in some cases memory on node-017 that cannot be allocated at all.

**Fix:** Set the default per partition from each node's real memory per CPU, with headroom for the operating system. Replace the cap with either the real ratio or memory-based billing.

**Advantages:** Nodes can be filled completely, with no stranded CPUs or memory. Users get the CPUs they asked for, and memory-heavy work is charged fairly for the share of the node it occupies.

**User advice:**

- Jobs that do not request memory with `--mem` or `--mem-per-cpu` will receive slightly less memory than before. For most jobs this makes no difference.
- If your job is memory sensitive, set its memory requirement explicitly in your submission script, for example `#SBATCH --mem=16G` or `#SBATCH --mem-per-cpu=4G`.
- Jobs already queued at the changeover keep the memory they were given at submission. The new defaults only apply to jobs submitted afterwards.

### Head node and login node upgrade to RHEL 9.8

<p>
  <span style="display:inline-block; background:#1f4e79; color:#fff; padding:3px 10px; border-radius:4px; font-weight:600;">25 November 2026</span>
  &nbsp; <strong>Downtime:</strong> 1 day
</p>

This change will upgrade the Anatra head node and login node from RHEL 9.6 to RHEL 9.8 to keep the cluster on a supported operating system version and to apply maintenance and security updates.

**Impact:** During the maintenance window, cluster access, job submission and management services will be unavailable. Running jobs on compute nodes may be interrupted, run with degraded performance or fail because of the temporary loss of communication with cluster management services. Post-upgrade validation and rollback procedures will be in place to minimise risk and restore service if required.

**User advice:**

- Save your work and log out of the login node before the maintenance window begins.
- Avoid submitting jobs whose run time could overlap the maintenance window, as they may fail. Where possible, set a time limit with `--time` so that jobs finish before the window starts, or hold them until the upgrade is complete.
- Jobs still queued when the window starts will remain in the queue and start once the service is restored.
