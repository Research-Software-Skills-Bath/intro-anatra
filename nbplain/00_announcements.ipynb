## Anatra maintenance programme, November to December 2026

We are carrying out a series of changes to Anatra to bring the operating system up to date and to improve job scheduling and resource utilisation across the cluster. The work starts in the week of 16 November 2026 and will continue until the end of the year.

Where a change causes downtime we will notify all researchers using Anatra by email with approximately two weeks' notice.

The schedule below covers the changes confirmed so far. We will keep it updated as further approvals are put in place, so we advise you to revisit this page for the latest information.

### 16 November 2026: Adjusting DefMemPerCPU and MaxMemPerCPU

**Issue:** Every partition currently uses the same memory-per-CPU default and cap (5,304 MB), which matches none of the hardware. Jobs that do not request memory leave CPUs idle on some nodes, memory-heavy jobs are silently allocated extra CPUs they do not use, and about 278 GiB on node017 can never be allocated.

**Fix:** Set the default per partition from each node's real memory per CPU, with headroom for the operating system. Replace the cap with either the real ratio or memory-based billing.

**Advantages:** Nodes can be filled completely, with no stranded CPUs or memory. Users get the CPUs they asked for, and memory-heavy work is charged fairly for the share of the node it occupies.

### 25 November 2026: Head node and login node upgrade to RHEL 9.8

This change will upgrade the Anatra head node and login node from RHEL 9.6 to RHEL 9.8 to keep the cluster on a supported operating system version and to apply maintenance and security updates.

**Downtime: 1 day.** During the maintenance window, cluster access, job submission and management services will be unavailable. Running jobs on compute nodes may be interrupted, run with degraded performance or fail because of the temporary loss of communication with cluster management services. Post-upgrade validation and rollback procedures will be in place to minimise risk and restore service if required.
